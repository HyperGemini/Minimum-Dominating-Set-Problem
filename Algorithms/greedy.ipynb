```python
def dominating_set_greedy(graph):
    graph_nodes = set(graph.keys())
    dominated_nodes = set()
    dominating_set = set()

    while dominated_nodes != graph_nodes:
        best_node = None
        best_coverage = 0

        for node in graph_nodes:
            covered_nodes = set(graph[node])
            covered_nodes.add(node)

            new_coverage = len(covered_nodes.difference(dominated_nodes))

            if new_coverage > best_coverage:
                best_coverage = new_coverage
                best_node = node

        dominating_set.add(best_node)
        dominated_nodes.add(best_node)
        dominated_nodes.update(graph[best_node])

    return dominating_set
```

```python
g = {
    1: [2, 5],
    2: [1, 3, 5],
    3: [2, 4, 6],
    4: [3],
    5: [1, 2, 6],
    6: [3, 5],
    7: []
}

print(dominating_set_greedy(g))
```
