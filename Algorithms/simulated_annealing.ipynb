```python
import random
from copy import deepcopy
```

```python
def is_dominating_set(solution_nodes, graph):
    all_nodes = set(graph.keys())
    for node in all_nodes:
        is_dominated = False
        if node in solution_nodes:
            is_dominated = True
        else:
            for neighbor in graph[node]:
                if neighbor in solution_nodes:
                    is_dominated = True
                    break
        if not is_dominated:
            return False
    return True
```

```python
def initialize(graph):
    return set(graph.keys())
```

```python
def calculate_solution_value(solution_nodes):
    return len(solution_nodes)
```

```python
def make_small_change(solution_nodes, graph):
    new_solution_nodes = deepcopy(solution_nodes)
    move_type = random.choice(['removal', 'swap'])

    if move_type == 'removal':
        if not new_solution_nodes:
            return new_solution_nodes
        node_to_remove = random.choice(list(new_solution_nodes))
        new_solution_nodes.remove(node_to_remove)

        if is_dominating_set(new_solution_nodes, graph):
            return new_solution_nodes
        else:
            return solution_nodes
    else:
        if not new_solution_nodes:
            return solution_nodes
        node_to_remove = random.choice(list(new_solution_nodes))
        new_solution_nodes.remove(node_to_remove)

        candidates = list(set(graph.keys()).difference(new_solution_nodes))
        if not candidates:
            return solution_nodes
        node_to_add = random.choice(candidates)
        new_solution_nodes.add(node_to_add)

        if is_dominating_set(new_solution_nodes, graph):
            return new_solution_nodes
        else:
            return solution_nodes
```

```python
def simulated_annealing(graph, max_iters):
    solution_nodes = initialize(graph)
    value = calculate_solution_value(solution_nodes)
    best_solution_nodes = deepcopy(solution_nodes)
    best_value = value

    for i in range(1, max_iters + 1):
        new_solution_nodes = make_small_change(solution_nodes, graph)
        new_value = calculate_solution_value(new_solution_nodes)

        if new_value < value:
            value = new_value
            solution_nodes = deepcopy(new_solution_nodes)
            if new_value < best_value:
                best_value = new_value
                best_solution_nodes = deepcopy(new_solution_nodes)
            else:
                p = 1 / (i ** 0.5)
                q = random.random()
                if q < p:
                    value = new_value
                    solution_nodes = deepcopy(new_solution_nodes)

    return best_solution_nodes, best_value
```

```python
import networkx as nx
```

```python
if __name__ == '__main__':
    g = {
        1: [2, 5],
        2: [1, 3, 5],
        3: [2, 4, 6],
        4: [3],
        5: [1, 2, 6],
        6: [3, 5]
    }

    max_iterations = 100

    best_solution, best_solution_size = simulated_annealing(g, max_iterations)

    print(f"Minimum dominating set: {best_solution}")
    print(f"Size of minimum dominating set: {best_solution_size}")

    # Provera
    g_graph_nx = nx.Graph(g)
    is_dominating = nx.is_dominating_set(g_graph_nx, best_solution)

    if is_dominating:
        print(f"{best_solution} is dominating set.")
```
